# Validation Plots (Editable)

This notebook is arranged so each figure has its own plotting cell.
The plotting code is expanded directly inside each figure cell instead of calling one shared plotting function.

Usage:
1. Run the setup cells first.
2. Open the figure cell you want.
3. Edit its local parameters at the top of that cell, such as `figsize`, `annotation_xy`, `x_limits`, `y_limits`, and legend positions.

```python
from src.utils import read_from_excel, resolve_lbl_reference

def maybe_save(fig: plt.Figure, stem: str | None = None) -> None:
    fig.tight_layout()
    if SAVE_FIGURES and stem:
        OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
        png_path = OUTPUT_DIR / f"{stem}.png"
        pdf_path = OUTPUT_DIR / f"{stem}.pdf"
        fig.savefig(png_path, dpi=gvp.FIGURE_DPI, bbox_inches="tight")
        fig.savefig(pdf_path, bbox_inches="tight")
        print(f"Saved {png_path}")
        print(f"Saved {pdf_path}")
    plt.show()


CASE_RESULTS = {}


def get_case_result(case_name: str, config_index: int):
    cache_key = (case_name, config_index)
    if cache_key in CASE_RESULTS:
        return CASE_RESULTS[cache_key]

    file_path = PROJECT_ROOT / "input" / "profiles" / f"{case_name}.npy"
    profile = np.load(file_path, allow_pickle=False)
    # reference_results = np.load(PROJECT_ROOT / "results" / f"{case_name}.npy") / 1e3
    reference_results = np.load(resolve_lbl_reference(f"{case_name}.npy")) / 1e3

    xs = profile["x"]
    length_value = float(case_name.split("_")[1][1:-1])
    x_rel = xs / length_value if length_value > 0 else xs
    temperature_profile = profile["T"]
    x_h2o = profile["Y_H2O"]
    x_nh3 = profile["Y_NH3"] + 1e-4

    wang_q, wang_dq = gvp.wsgg_wang_case_solution(xs, temperature_profile, x_h2o, x_nh3, WANG_FILE, SHEET_NAME, BOUNDARY_TEMPERATURE)
    rsm_q, rsm_dq = gvp.wsgg_rsm_case_solution(xs, temperature_profile, x_h2o, x_nh3, beta, gamma, T_REF, BOUNDARY_TEMPERATURE)

    wang_q = wang_q / 1e3
    wang_dq = wang_dq / 1e3
    rsm_q = rsm_q / 1e3
    rsm_dq = rsm_dq / 1e3

    reference_q = reference_results[:, 1]
    reference_st = -reference_results[:, 2]
    delta_q_rsm = gvp.relative_error_percent(reference_q, rsm_q)
    delta_q_wang = gvp.relative_error_percent(reference_q, wang_q)
    delta_st_rsm = gvp.relative_error_percent(reference_st, -rsm_dq)
    delta_st_wang = gvp.relative_error_percent(reference_st, -wang_dq)
    error_limit = gvp.rounded_error_limit(delta_q_rsm, delta_q_wang, delta_st_rsm, delta_st_wang)

    result = {
        "case_name": case_name,
        "config_index": config_index,
        "config_label": f"Cfg{config_index}",
        "length_value": length_value,
        "x_rel": x_rel,
        "reference_q": reference_q,
        "reference_st": reference_st,
        "wang_q": wang_q,
        "wang_st": -wang_dq,
        "rsm_q": rsm_q,
        "rsm_st": -rsm_dq,
        "delta_q_rsm": delta_q_rsm,
        "delta_q_wang": delta_q_wang,
        "delta_st_rsm": delta_st_rsm,
        "delta_st_wang": delta_st_wang,
        "error_limit": error_limit,
    }
    CASE_RESULTS[cache_key] = result
    return result
```

```python
# ============================================================================
# Comment 8 helpers - Wang et al. validity envelope & extrapolation marking
#   Defined HERE (not in src/generate_validation_plots.py) so that the
#   underlying calculation logic is left untouched.
# ============================================================================

# Stated validity range of the Wang et al. WSGG model for H2O/NH3
# (Px = 0.2 bar sheet):  300 <= T <= 2300 K,  0.1 <= L <= 20 m
WANG_VALIDITY = {
    "T_min": 300.0,
    "T_max": 2300.0,
    "L_min": 0.1,
    "L_max": 20.0,
}

# master switch: set False to reproduce the original (unmarked) figures
MARK_EXTRAPOLATION = True


def mark_extrapolation(ax, x_values, limit: float, limit_low: float | None = None) -> bool:
    """Shade the part of the x axis that lies OUTSIDE the comparison model's range.

    A light-grey band (``axvspan``) covers ``x > limit`` (and, optionally,
    ``x < limit_low``) and a dashed vertical line marks the boundary.
    Returns True when at least one band was drawn, so the caller can decide
    whether to add the corresponding legend entry.
    """
    x_values = np.asarray(x_values, dtype=float)
    drawn = False

    x_max = float(np.max(x_values))
    if x_max > limit:
        ax.axvspan(limit, x_max, color="#EDEDED", zorder=0, linewidth=0.0)
        ax.axvline(limit, color="#9E9E9E", linewidth=0.8,
                   linestyle=(0, (4, 2)), zorder=1)
        drawn = True

    if limit_low is not None:
        x_min = float(np.min(x_values))
        if x_min < limit_low:
            ax.axvspan(x_min, limit_low, color="#EDEDED", zorder=0, linewidth=0.0)
            ax.axvline(limit_low, color="#9E9E9E", linewidth=0.8,
                       linestyle=(0, (4, 2)), zorder=1)
            drawn = True

    return drawn


def wang_inside_mask(reference_shape, temperatures, path_lengths) -> np.ndarray:
    """Boolean mask (flattened, same order as ``reference.ravel()``) that is True
    for the points lying inside the Wang et al. validity envelope."""
    t_grid, l_grid = np.meshgrid(np.asarray(temperatures, dtype=float),
                                 np.asarray(path_lengths, dtype=float),
                                 indexing="ij")
    envelope = (
        (t_grid >= WANG_VALIDITY["T_min"]) & (t_grid <= WANG_VALIDITY["T_max"])
        & (l_grid >= WANG_VALIDITY["L_min"]) & (l_grid <= WANG_VALIDITY["L_max"])
    )
    return np.broadcast_to(envelope, reference_shape).ravel()


def stats_text_with_n(label: str, reference: np.ndarray, prediction: np.ndarray) -> str:
    """``gvp.stats_text`` with the sample count inserted on the second line."""
    lines = gvp.stats_text(label, reference, prediction).split("\n")
    return "\n".join([lines[0], f"$N$ = {len(reference):,}"] + lines[1:])


print("Wang et al. envelope :", WANG_VALIDITY)
print("Mark extrapolation   :", MARK_EXTRAPOLATION)

```

## Parity Plots


```python
# Editable parameters for this figure
figsize = (5.5, 5.0)
x_limits = (0.0, 1.0)
y_limits = (0.0, 1.0)
stats_xy = (0.98, 0.04)
guide_legend_loc = "upper left"
save_stem = "parity_rsm_kde"

x = val_rsm.ravel()
y = val_ref.ravel()
xy = np.vstack([x, y])
density = gaussian_kde(xy)(xy)
order = np.argsort(density)

fig, ax = plt.subplots(figsize=figsize)
gvp.draw_parity_guides(ax, x_limits)

scatter = ax.scatter(
    x[order],
    y[order],
    c=density[order],
    cmap="cividis",
    s=10,
    linewidths=0.0,
    alpha=0.9,
    rasterized=True,
    zorder=2,
)
gvp.style_axes(ax)
ax.set_xlim(*x_limits)
ax.set_ylim(*y_limits)
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel("Predicted total emissivity")
ax.set_ylabel("Reference total emissivity (LBL)")

# 图例保持不变
ax.legend(handles=gvp.parity_guide_handles(), loc=guide_legend_loc)

# ✅ 统计框：Present RSM-WSGGM 一行，N 一行，再是 R²/RMSE
stats_base = gvp.stats_text("Present RSM-WSGGM", y, x)
lines = stats_base.split("\n")
new_lines = [lines[0], f"N = {len(x):,}"] + lines[1:]
stats_text = "\n".join(new_lines)

gvp.annotate_box(
    ax,
    stats_text,
    stats_xy,
    ha="right",
    va="bottom"
)

colorbar = fig.colorbar(scatter, ax=ax, fraction=0.05, pad=0.02)
colorbar.set_label("Point density")
colorbar.ax.tick_params(direction="in")

maybe_save(fig, save_stem if SAVE_FIGURES else None)
```

```python
# Editable parameters for this figure
figsize = (5.0, 4.6)
x_limits = (0.0, 1.0)
y_limits = (0.0, 1.0)
guide_legend_loc = "upper left"

# Comment 8: the statistics box is split into three, so that the advantage of the
# present model can be read separately from the region where the Wang et al.
# model is being extrapolated beyond its stated validity envelope.
SPLIT_STATS_BOXES = True
MARK_EXTRAPOLATED_POINTS = True          # draw out-of-envelope Wang points as grey squares
# The boxes sit in the two triangles that the parity cloud leaves empty:
# the present-model box on the left (x < y), the two Wang boxes on the right (x > y).
# stats_xy = {
#     "present":      (0.02, 0.42),    # ha="left",  va="bottom"
#     "wang_inside":  (0.99, 0.27),    # ha="right", va="bottom"
#     "wang_extrap":  (0.99, 0.03),    # ha="right", va="bottom"
# }

stats_xy = {
    "present":      (0.53, 0.03),    # ha="left",  va="bottom"
    "wang_inside":  (0.99, 0.29),    # ha="right", va="bottom"
    "wang_extrap":  (0.99, 0.03),    # ha="right", va="bottom"
}
save_stem = "parity_models_vs_lbl"

y = val_ref.ravel()
x_rsm = val_rsm.ravel()
x_wang = val_wang.ravel()

# --- Comment 8: split the Wang points by the stated validity envelope --------
inside = wang_inside_mask(val_ref.shape, val_t, val_l)
n_inside = int(np.count_nonzero(inside))
n_outside = int(inside.size - n_inside)
print(f"Wang points inside envelope : {n_inside:,}")
print(f"Wang points extrapolated    : {n_outside:,}")

fig, ax = plt.subplots(figsize=figsize)
gvp.draw_parity_guides(ax, x_limits)

# ±10 % guide lines (kept from the original cell)
ax.plot(x_limits, [1.1 * v for v in x_limits], color="#808080", linestyle="--",
        linewidth=0.8, alpha=0.6, zorder=1)
ax.plot(x_limits, [0.9 * v for v in x_limits], color="#808080", linestyle="--",
        linewidth=0.8, alpha=0.6, zorder=1)

# Wang et al. - points outside the stated envelope (grey hollow squares)
if SPLIT_STATS_BOXES and MARK_EXTRAPOLATED_POINTS and n_outside:
    ax.scatter(x_wang[~inside], y[~inside], s=11, marker="s", facecolors="none",
               edgecolors="#BDBDBD", linewidths=0.45, alpha=0.5,
               rasterized=True, zorder=2)

# Wang et al. - inside the envelope
ax.scatter(x_wang[inside] if SPLIT_STATS_BOXES else x_wang,
           y[inside] if SPLIT_STATS_BOXES else y,
           s=10, marker="*", facecolors="none", edgecolors=gvp.COLORS["wang"],
           linewidths=0.45, alpha=0.35, rasterized=True, zorder=3)

# Present model (all points)
ax.scatter(x_rsm, y, s=10, marker="o", facecolors="none",
           edgecolors=gvp.COLORS["present"], linewidths=0.45, alpha=0.35,
           rasterized=True, zorder=4)



gvp.style_axes(ax)
ax.set_xlim(*x_limits)
ax.set_ylim(*y_limits)
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel("Predicted total emissivity")
ax.set_ylabel("LBL total emissivity")

# --- single combined legend (guides + markers) -------------------------------
handles = gvp.parity_guide_handles() + [
    Line2D([0], [0], linestyle="None", marker="o", markerfacecolor="none",
           markeredgecolor=gvp.COLORS["present"], markeredgewidth=0.9,
           markersize=5.0, label="Present model"),
    Line2D([0], [0], linestyle="None", marker="*", markerfacecolor="none",
           markeredgecolor=gvp.COLORS["wang"], markeredgewidth=0.9,
           markersize=6.0,
           label="Wang et al. WSGGM (inside range)" if SPLIT_STATS_BOXES else "Wang et al. WSGGM"),
]
if SPLIT_STATS_BOXES and MARK_EXTRAPOLATED_POINTS and n_outside:
    handles.append(
        Line2D([0], [0], linestyle="None", marker="s", markerfacecolor="none",
               markeredgecolor="#BDBDBD", markeredgewidth=0.9, markersize=5.0,
               label="Wang et al. WSGGM (extrapolated)")
    )
ax.legend(handles=handles, loc=guide_legend_loc)

# --- statistics boxes ---------------------------------------------------------
if SPLIT_STATS_BOXES:
    gvp.annotate_box(ax, stats_text_with_n("Present (all)", y, x_rsm),
                     stats_xy["present"], ha="left", va="bottom")
    gvp.annotate_box(ax, stats_text_with_n("Wang (inside range)", y[inside], x_wang[inside]),
                     stats_xy["wang_inside"], ha="right", va="bottom")
    if n_outside:
        gvp.annotate_box(ax, stats_text_with_n("Wang (extrapolated)", y[~inside], x_wang[~inside]),
                         stats_xy["wang_extrap"], ha="right", va="bottom")
else:
    gvp.annotate_box(ax, stats_text_with_n("Present", y, x_rsm), stats_xy["present"],
                     ha="left", va="bottom")
    gvp.annotate_box(ax, stats_text_with_n("Wang", y, x_wang), stats_xy["wang_inside"],
                     ha="right", va="bottom")



maybe_save(fig, save_stem if SAVE_FIGURES else None)

```

## Emissivity vs Temperature


```python
# Editable parameters for this figure
figsize = (4.5, 4.9)
selected_paths = [0.2, 1.0, 5.0, 10.0, 30.0, 60.0]
annotation_xy = (0.97, 0.96)
series_legend_loc = "lower left"          # Comment 8: was "upper left"
series_legend_anchor = (0.0, 0.0)         # Comment 8: was (0.55, 0.89)
series_legend_ncol = 2
model_legend_loc = "upper right"          # Comment 8: unchanged from original
model_legend_anchor = (0.98, 0.98)          # Comment 8: new
x_limits = None
y_limits = None
save_stem = "emissivity_vs_temperature_mr_0p500"
# Comment 8: no horizontal shading here. The extrapolation condition for this
# figure is P_t*L (not T), so it is flagged per curve instead (see below).
extrapolation_limit = WANG_VALIDITY["T_max"]

mr_value = 0.5
temperature_order = np.argsort(main_t)
path_order = np.argsort(main_l)
main_l_sorted = main_l[path_order]
mr_index = gvp.nearest_index(main_mr, mr_value)
selected_path_indices = [gvp.nearest_index(main_l_sorted, path) for path in selected_paths]

x_values = main_t[temperature_order]
reference_curves = [main_ref[mr_index, temperature_order, idx] for idx in selected_path_indices]
rsm_curves = [main_rsm[mr_index, temperature_order, idx] for idx in selected_path_indices]
wang_curves = [main_wang[mr_index, temperature_order, idx] for idx in selected_path_indices]
series_labels = ["{:.1f}".format(main_l_sorted[idx]) for idx in selected_path_indices]
annotation_text = "$M_r = {:.2f}$".format(mr_value)

# Comment 8: P_t*L > 20 bar m lies outside the Wang et al. range, so the whole
# curve is drawn dashed. This is where the comparison is being extrapolated.
path_values = [main_l_sorted[idx] for idx in selected_path_indices]
path_linestyles = ["--" if v > WANG_VALIDITY["L_max"] else "-" for v in path_values]

if y_limits is None:
    y_limits = gvp.emissivity_ylim(reference_curves, rsm_curves, wang_curves)
if x_limits is None:
    x_limits = (float(x_values.min()), float(x_values.max()))

colors = gvp.choose_series_colors(len(series_labels))
fig, ax = plt.subplots(figsize=figsize)

# Comment 8: intentionally no axvspan. The temperature axis only reaches
# ~2348 K against a 2300 K limit, so the shaded band would be ~2% of the axis.
if MARK_EXTRAPOLATION:
    mark_extrapolation(ax, x_values, extrapolation_limit)

for color, reference_curve, rsm_curve, wang_curve, line_style in zip(
        colors, reference_curves, rsm_curves, wang_curves, path_linestyles):
    ax.plot(x_values, reference_curve, color=color, linewidth=1.45,
            zorder=1, linestyle=line_style)
    ax.scatter(
        x_values,
        rsm_curve,
        marker="o",
        s=15,
        facecolors="white",
        edgecolors=color,
        linewidths=1.0,
        zorder=3,
    )
    ax.scatter(
        x_values,
        wang_curve,
        marker="*",
        s=14,
        facecolors="white",
        edgecolors=color,
        linewidths=0.95,
        zorder=3,
    )

gvp.style_axes(ax, grid_axis="y", logx=False)
ax.set_xlim(*x_limits)
ax.set_ylim(*y_limits)
ax.set_xlabel(r"Temperature, $T$ (K)")
ax.set_ylabel(r"Total emissivity, $\varepsilon$")
# gvp.annotate_box(ax, annotation_text, annotation_xy, ha="right")

series_handles = [
    Line2D([0], [0], color=color, linewidth=1.6, linestyle=line_style, label=label)
    for color, label, line_style in zip(colors, series_labels, path_linestyles)
]
legend_series = ax.legend(
    handles=series_handles,
    title=r"$P_tL$ (bar m)",
    loc=series_legend_loc,
    bbox_to_anchor=series_legend_anchor,
    borderaxespad=0.0,
    ncol=series_legend_ncol,
    columnspacing=0.8,
    handlelength=1.8,
    frameon=False,
    fontsize=8,
)

model_handles = [
    Line2D(
        [0], [0],
        linestyle="None",
        marker="o",
        markerfacecolor="white",
        markeredgecolor="black",
        markeredgewidth=1.0,
        markersize=5.0,
        label="Present model",
    ),
    Line2D(
        [0], [0],
        linestyle="None",
        marker="*",
        markerfacecolor="white",
        markeredgecolor="black",
        markeredgewidth=1.0,
        markersize=4.8,
        label="Wang et al. WSGGM",
    ),
]
# Comment 8: same wording as the other figures - dashed marks the extrapolated region
model_handles = model_handles + [
    Line2D([0], [0], color="#555555", linewidth=1.4, linestyle="--",
           label="Dashed: outside Wang range"),
]
legend_models = ax.legend(
    handles=model_handles,
    loc=model_legend_loc,
    bbox_to_anchor=model_legend_anchor,
    borderaxespad=0.0,
    ncol=1,
    frameon=False,
    fontsize=8,
)

ax.add_artist(legend_series)
ax.add_artist(legend_models)

maybe_save(fig, save_stem if SAVE_FIGURES else None)
```

Reuses `get_case_result()` so the numbers are identical to those plotted above.
All errors are normalised percentages: `*_max` = max |error|, `*_avg` = mean |error|.



```python
# ============================================================================
# Error statistics table for all cases  (LaTeX in the style of Table 3)
# ============================================================================
# delta_q_* / delta_st_* returned by get_case_result() are already normalised
# percentages, i.e. (model - LBL) / max|LBL| * 100:
#   dq_max = max|delta_q|   dq_avg = mean|delta_q|
#   dS_max = max|delta_st|  dS_avg = mean|delta_st|

import pandas as pd

rows = []
for index, path in enumerate(CASE_FILES, start=1):
    stem = path.stem
    label = stem.split("_")[0].replace("Case", "Case ")      # Case1.1 -> "Case 1.1"
    try:
        res = get_case_result(stem, index)
    except Exception as exc:                                  # missing LBL, Mr out of range, ...
        print(f"[skip] {stem}: {type(exc).__name__}: {exc}")
        continue
    rows.append({
        "Case":    label,
        "L (m)":   res["length_value"],
        "W dq_max": float(np.max(np.abs(res["delta_q_wang"]))),
        "W dq_avg": float(np.mean(np.abs(res["delta_q_wang"]))),
        "W dS_max": float(np.max(np.abs(res["delta_st_wang"]))),
        "W dS_avg": float(np.mean(np.abs(res["delta_st_wang"]))),
        "P dq_max": float(np.max(np.abs(res["delta_q_rsm"]))),
        "P dq_avg": float(np.mean(np.abs(res["delta_q_rsm"]))),
        "P dS_max": float(np.max(np.abs(res["delta_st_rsm"]))),
        "P dS_avg": float(np.mean(np.abs(res["delta_st_rsm"]))),
    })

df_errors = pd.DataFrame(rows)

if df_errors.empty:
    print("No case results available - check input/profiles and results/lbl.")
else:
    # ---------------- display ----------------
    display(
        df_errors.style
        .format({
            "L (m)": "{:.1f}",
            "W dq_max": "{:.2f}", "W dq_avg": "{:.2f}",
            "W dS_max": "{:.2f}", "W dS_avg": "{:.2f}",
            "P dq_max": "{:.2f}", "P dq_avg": "{:.2f}",
            "P dS_max": "{:.2f}", "P dS_avg": "{:.2f}",
        })
        .background_gradient(cmap="OrRd", subset=[c for c in df_errors.columns if c not in ("Case", "L (m)")])
        .set_caption("Normalised errors (%) of the radiative heat flux and source term")
        .hide(axis="index")
    )

    # ---------------- LaTeX (Table 3 style) ----------------
    keys = ["W dq_max", "W dq_avg", "P dq_max", "P dq_avg",
            "W dS_max", "W dS_avg", "P dS_max", "P dS_avg"]

    # per-column width, left aligned (auto-adapts to the number of digits)
    widths  = [max(len(f"{r[k]:.2f}") for r in rows) for k in keys]
    label_w = max(len(r["Case"]) for r in rows)

    all_L1 = all(abs(r["L (m)"] - 1.0) < 1e-9 for r in rows)
    caption = (r"\caption{Normalized errors of the radiative heat flux and source term "
               r"for all cases at $P_t = 1$~bar and $L = 1$~m.}" if all_L1 else
               r"\caption{Normalized errors of the radiative heat flux and source term "
               r"for all cases at $P_t = 1$~bar.}")

    tex_lines = [
        r"\begin{table}[htbp]",
        r"\centering",
        caption,
        r"\label{tab:error_summary}",
        r"\begin{tabular}{lcccccccc}",
        r"\hline",
        r"\multirow{2}{*}{Errors (\%)} ",
        r" & \multicolumn{2}{c}{Wang et al.} ",
        r" & \multicolumn{2}{c}{Present} ",
        r" & \multicolumn{2}{c}{Wang et al.} ",
        r" & \multicolumn{2}{c}{Present} \\",
        r" \cline{2-9}",
        r" & $\delta q_{\max}$ & $\delta q_{\mathrm{avg}}$"
        r" & $\delta q_{\max}$ & $\delta q_{\mathrm{avg}}$"
        r" & $\delta \dot{q}_{\max}$ & $\delta \dot{q}_{\mathrm{avg}}$"
        r" & $\delta \dot{q}_{\max}$ & $\delta \dot{q}_{\mathrm{avg}}$ \\",
        r"\hline",
    ]
    for r in rows:
        cells = " & ".join(f"{r[k]:.2f}".ljust(w) for k, w in zip(keys, widths))
        tex_lines.append(f"{r['Case']:<{label_w}} & {cells} \\\\")
    tex_lines += [r"\hline", r"\end{tabular}", r"\end{table}"]

    tex_src = "\n".join(tex_lines)
    print(tex_src)

    # ---------------- save ----------------
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    csv_path = OUTPUT_DIR / "case_error_summary.csv"
    tex_path = OUTPUT_DIR / "case_error_summary.tex"
    df_errors.to_csv(csv_path, index=False, encoding="utf-8-sig")
    tex_path.write_text(tex_src + "\n", encoding="utf-8")
    print(f"\nSaved {csv_path}")
    print(f"Saved {tex_path}")
```
